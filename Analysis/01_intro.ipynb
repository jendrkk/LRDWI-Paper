```python
import pandas as pd
import numpy as np
from pathlib import Path
import os, sys, time

REPO = Path(os.getcwd()).parent
DATA_ROOT = Path(os.path.join(REPO, '..', '..', 'Data', 'replication_package'))
PLOT_ROOT = DATA_ROOT.parent.parent / 'Plots' / 'Country'

sys.path.insert(0, os.path.join(REPO, 'Code', 'tools'))
```

```python
# Load CBOS data
cbos = pd.read_csv(DATA_ROOT / "CBOS" / "CBOS_survey.csv", low_memory=False)
```

```python
cbos[['educ_2000', 'educ_2000_L']].drop_duplicates()
```

```python

# Descriptive statistics per year
desc_dict = {}

cbos_temp = cbos.copy()
cbos_temp['sex'] = cbos['sex'] -1

for year, grp in cbos_temp.groupby('survey_year'):
    desc_dict[year] = {
        ('N',              'Count'):   grp.shape[0],
        ('Sex',            '%'):   np.round(grp['sex'].mean() *100, 1),
        ('Age',            'P25'):    np.round(grp['age'].quantile(0.25),0),
        ('Age',            'P50'):    np.round(grp['age'].median(),0),
        ('Age',            'P75'):    np.round(grp['age'].quantile(0.75),0),
        ('Education_2',      '%'):   np.round(grp[grp['educ_2000'] == 2].shape[0] / grp.shape[0] * 100,1),
        ('Education_3',      '%'):   np.round(grp[grp['educ_2000'] == 3].shape[0] / grp.shape[0] * 100,1),
        ('Education_4',      '%'):   np.round(grp[grp['educ_2000'] == 4].shape[0] / grp.shape[0] * 100,1),
        ('Education_5',      '%'):   np.round(grp[grp['educ_2000'] == 5].shape[0] / grp.shape[0] * 100,1),
        ('Household size', 'P25'):    np.round(grp['household_size'].quantile(0.25),0),
        ('Household size', 'P50'):    np.round(grp['household_size'].median(),0),
        ('Household size', 'P75'):    np.round(grp['household_size'].quantile(0.75),0),
    }

desc_stats = pd.DataFrame(desc_dict).T
desc_stats.index.name = 'Year'
desc_stats.columns = pd.MultiIndex.from_tuples(desc_stats.columns)
desc_stats = desc_stats.round(2)

latex = desc_stats.to_latex(
    multicolumn=True,
    multicolumn_format='c',
    multirow=True,
    caption='Descriptive statistics by survey year',
    label='tab:desc_stats',
    position='h',
    float_format="%.1f"
)
print(latex)
```
